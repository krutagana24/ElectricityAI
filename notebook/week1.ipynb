```python
df = pd.read_csv("../data/electricity_sample.csv")
```

```python
df.shape
```

```python
import numpy as np

df.replace("?", np.nan, inplace=True)
```

```python
df.isnull().sum()
```

```python
numeric_columns = [
    "Global_active_power",
    "Global_reactive_power",
    "Voltage",
    "Global_intensity",
    "Sub_metering_1",
    "Sub_metering_2",
    "Sub_metering_3"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
df.dtypes
```

```python
df["Datetime"] = pd.to_datetime(
    df["Date"] + " " + df["Time"],
    format="%d/%m/%Y %H:%M:%S"
)
```

```python
df.head()
```

```python
df.drop(
    columns=["Date", "Time"],
    inplace=True
)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.shape
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.shape
```

```python
df.to_csv("../data/electricity_cleaned.csv", index=False)

print("Cleaned Dataset Saved Successfully!")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")

sns.set(font_scale=1.1)
```

```python
df.describe().T
```

```python
plt.figure(figsize=(12,8))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(
    df["Global_active_power"],
    bins=40,
    kde=True
)

plt.title("Distribution of Global Active Power")
plt.xlabel("Global Active Power")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(
    df["Voltage"],
    bins=40,
    color="green",
    kde=True
)

plt.title("Voltage Distribution")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(
    df["Global_intensity"],
    bins=40,
    color="orange",
    kde=True
)

plt.title("Global Intensity Distribution")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    x=df["Global_active_power"]
)

plt.title("Outlier Detection")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    x=df["Voltage"],
    color="lightblue"
)

plt.show()
```

```python
sample = df.sample(1000)

sns.pairplot(sample)

plt.show()
```

```python
plt.figure(figsize=(15,6))

plt.plot(
    df["Datetime"],
    df["Global_active_power"]
)

plt.title("Electricity Consumption Over Time")

plt.xlabel("Time")

plt.ylabel("Global Active Power")

plt.show()
```

```python
df["Month"] = df["Datetime"].dt.month

monthly = df.groupby("Month")["Global_active_power"].mean()

monthly.plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Average Monthly Power Consumption")

plt.show()
```
